```python
weights = np.random.randn(len(input), 3) * np.sqrt(2/3)
```

```python
def sigmoid(z):
    return 1/(1 + np.exp(-np.clip(z, -100, 100)))
```

```python
bias = np.zeros((1, 3))
```

```python
z = np.dot(input, weights) + bias
```

```python
z
```

```python
a = sigmoid(z)
```

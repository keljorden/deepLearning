TORCH WorkFlow

```python
import torch
from torch import nn
import matplotlib.pyplot as plt
```

```python
torch.__version__
```

```python
#creatin synthatic data using linear regression
w = 0.75
b = 0.25

start = -1
end = 1
steps = 0.02

X = torch.arange(start, end, steps).unsqueeze(dim = 1)
y = w*X + b

X[:10], y[:10]
```

```python
plt.plot(X,y)
```

```python
train_split = int(0.8 * len(X))

X_train, y_train = X[:train_split], y[:train_split]
X_test, y_test = X[train_split:], y[train_split:]

len(X_train), len(X_test), len(y_train), len(y_test)
```

```python
def plot_predictions(train_data=X_train, train_labels=y_train, test_data=X_test, test_labels=y_test, predictions=None):

  plt.figure(figsize=(10, 7))

  plt.scatter(train_data, train_labels, c="b", s=4, label="Training data")
  
  plt.scatter(test_data, test_labels, c="g", s=4, label="Testing data")

  if predictions is not None:
    plt.scatter(test_data, predictions, c="r", s=4, label="Predictions")

  plt.legend(prop={"size": 14})
```

```python
plot_predictions()
```

```python
class LinearRegression(nn.Module):
    def __init__ (self):
        super().__init__()

        self.weight = nn.Parameter(torch.rand(1, dtype = torch.float), requires_grad = True)
        self.bias = nn.Parameter(torch.rand(1, dtype = torch.float), requires_grad = True)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.weight * x + self.bias
```

```python
torch.manual_seed(42)

lr_model = LinearRegression()
list(lr_model.parameters())
```

```python
lr_model.state_dict()
```

```python
with torch.inference_mode():
    y_pred = lr_model(X_test)
```

```python
print(f"Number of testing samples: {len(X_test)}") 
print(f"Number of predictions made: {len(y_pred)}")
print(f"Predicted values:\n{y_pred}")
```

```python
y_pred.shape
```

```python
plot_predictions(predictions=y_pred)
```

```python
loss = y_pred - y_test
loss
```

```python
loss_fn = nn.L1Loss()

optimizer = torch.optim.SGD(params=lr_model.parameters(), lr=0.01)
```

```python
torch.manual_seed(42)
epochs = 100

train_loss_values = []
test_loss_values = []
epoch_count = []

for epoch in range(epochs):

    lr_model.train()

    y_pred = lr_model(X_train)

    loss = loss_fn(y_pred, y_train)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    lr_model.eval()

    with torch.inference_mode():
      test_pred = lr_model(X_test)

      test_loss = loss_fn(test_pred, y_test.type(torch.float)) 

      if epoch % 10 == 0:
            epoch_count.append(epoch)
            train_loss_values.append(loss.detach().numpy())
            test_loss_values.append(test_loss.detach().numpy())
            print(f"Epoch: {epoch} | MAE Train Loss: {loss:.4f} | MAE Test Loss: {test_loss:.4f} ")
```

```python
plt.plot(epoch_count, train_loss_values, label="Train loss")
plt.plot(epoch_count, test_loss_values, label="Test loss")
plt.title("Training and test loss curves")
plt.ylabel("Loss")
plt.xlabel("Epochs")
plt.legend()
```

```python
print("the original values for weights and bias are:")
print(f"weights: {w}, bias: {b}")
print("\nThe model learned the following values for weights and bias:")
print(lr_model.state_dict())
```

```python
from pathlib import Path

MODEL_PATH = Path("models")
MODEL_PATH.mkdir(parents=True, exist_ok=True)

MODEL_NAME = "lr_model.pth"
MODEL_SAVE_PATH = MODEL_PATH / MODEL_NAME

print(f"Saving model to: {MODEL_SAVE_PATH}")
torch.save(obj = lr_model.state_dict(), f=MODEL_SAVE_PATH)
```

```python
class LinearModel(nn.Module):
    def __init__(self):
        super().__init__()

        self.linear_layer = torch.nn.Linear(in_features = 1, out_features = 1)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.linear_layer(x)
```

```python
torch.manual_seed(42)
linear_model = LinearModel()
linear_model.state_dict()
```

```python
next(linear_model.parameters()).device
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
```

```python
linear_model.to(device)
next(linear_model.parameters()).device
```

```python
lf = torch.nn.L1Loss()
optimizer = torch.optim.SGD(params = linear_model.parameters(), lr = 0.01)
```

```python
torch.manual_seed(42)
epoch = 1000

X_train = X_train.to(device)
X_test = X_test.to(device)
y_train = y_train.to(device)
y_test = y_test.to(device)

for ep in range(epoch):

    linear_model.train()

    y_pred = linear_model(X_train)

    loss = lf(y_pred, y_train)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    linear_model.eval()
    with torch.inference_mode():

        t_loss = linear_model(X_test)
        tl = lf(t_loss, y_test)

        if ep % 50 == 0:
            print(f'epoch:\t{ep}\t||\tMAE train loss:\t{loss}\t||\tMAE test loss:\t{tl}')
```

```python
print("The model learned the following values for weights and bias:")
print(linear_model.state_dict())
print("\nAnd the original values for weights and bias are:")
print(f"weights: {w}, bias: {b}")
```

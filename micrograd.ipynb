```python
h = 1*10**-5
x = 3
print(h)
print(x)
print(f(x))
print(f(x + h))
```

```python
rise = (f(x + h) - f(x))
slop = rise/h

print(f'rise (increase in val of y):\t {rise}')
print(f'slop:\t {slop}')
```

```python
h = 1*10**-5
x = -3
print(f'h:\t{h}')
print(f'x:\t {x}')
print(f'f(x):\t {f(x)}')
print(f'f(x + h):\t {f(x+h)}')
```

```python
r = (f(x + h) - f(x))
s = r/h

print(f'rise (increase in val of y):\t {r}')
print(f'slop:\t {s}')
```

```python
h = 1 * 10 ** -5

a = 10
b = -3
c = 6

d1 = a * b + c
d2 = (a + h) * b + c

print('rise: ', d2 - d1)
print('slop: ', (d2-d1)/h)
```

```python
d3 = a * (b + h) + c

print('rise: ', d3 - d1)
print('slop: ', (d3-d1)/h)
```

```python
d4 = a * b + (c + h)

print('rise: ', d4 - d1)
print('slop: ', (d4-d1)/h)
```
